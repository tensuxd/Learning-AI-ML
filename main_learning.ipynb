```python
import numpy as np
import pandas as pd

data = pd.Series([10, 20, 30, 40, 50])

mean = np.mean(data)
median = np.median(data)
iqr = np.percentile(data, 75) - np.percentile(data, 25)

print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"IQR: {iqr}")
```
